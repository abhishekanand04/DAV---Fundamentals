```python
# 10 coin toss -> 7 are heads

from scipy.stats import binom

1 - binom.cdf(k=6, n=10, p=0.5)
```

```python
# alpha = 0.05 -> 95% confidence

alpha = 0.05
p = 1 - binom.cdf(k=6, n=10, p=0.5)
if (p < alpha):
  print("Reject H0")
else:
  print("Fail to reject H0")
```
